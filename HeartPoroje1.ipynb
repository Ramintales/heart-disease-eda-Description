```python


table2.loc[table2["age"] < 40, "age_group"] = "age<40"
table2.loc[(table2["age"] >= 40) & (table2["age"] < 60), "age_group"] = "40<age<59"
table2.loc[table2["age"] >= 60, "age_group"] = "Age>60"
```

## 18. Reviewing the Cleaned Dataset

The cleaned DataFrame is displayed again to verify the structure and values after the initial preprocessing steps.

At this point, duplicate records have been removed and the age-group feature has been added for further exploratory analysis.

```python


columns = ['chol', 'trestbps', 'thalach', 'oldpeak']

for col in columns:
    
    Q1 = table2[col].quantile(0.25)
    Q3 = table2[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = table2[(table2[col] < lower) | (table2[col] > upper)]

    print(f'Column: {col}')
    print(f'Number of Outliers: {len(outliers)}')
    print('----------------------')
```

## 19. Detecting Outliers Using the IQR Method

The Interquartile Range (IQR) method is used to identify potential outliers in important numerical variables.

The analysis focuses on cholesterol, resting blood pressure, maximum heart rate, and oldpeak. Observations outside the range of 1.5 times the IQR below Q1 or above Q3 are considered potential outliers.

This step helps identify unusual measurements that may require further investigation.

```python


plt.figure(figsize=(6,4))
sns.boxplot(x=table2["chol"])
plt.title('chol Boxplot')
plt.show()
```
